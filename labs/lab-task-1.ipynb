# Laboratory Task 1: Four Types of Data Analytics

<h3 style="font-family: Times New Roman"><strong>Laboratory Task 1</strong></h3>

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>Instruction:</b> From the example scenario (Dr. Jake, CEO of Astro JR, and the production and deployment of the COVID19 vaccine), create additional two inquiries for each type of data analytics.</p>

<div style="font-family: Times New Roman; font-size: 15px; text-align:justify">

<ol>

<li><b>Descriptive</b> <i>(What happened?)</i>
    <ol type="a">
        <li>How many doses of AstroJR's COVID19 vaccine were produced and distributed in each country during the past month?</li>
        <li>What percentage of vaccinated individuals in each region completed the full dose series over the last quarter?</li>
    </ol>
</li>
<br>
<li><b>Diagnostic</b> <i>(Why did it happen?)</i>
    <ol type="a">
        <li>Why did vaccine uptake drop in certain regions last month (e.g., supply delays, public hesitancy, or reports of side effects)?</li>
        <li>What factors caused delays in delivering vaccines to remote and island areas?</li>
    </ol>
</li>
<br>
<li><b>Predictive</b> <i>(What will happen?)</i>
    <ol type="a">
        <li>How many vaccine doses will each region need over the next month?</li>
        <li>How likely is a new surge of infections next quarter, given current vaccination rates and emerging variants?</li>
    </ol>
</li>
<br>
<li><b>Prescriptive</b> <i>(What should we do?)</i>
    <ol type="a">
        <li>If vaccine supply is limited, how should doses be allocated across regions to minimize the number of deaths?</li>
        <li>What production schedule and cold-chain distribution plan should AstroJR adopt to meet the forecasted demand at the lowest cost?</li>
    </ol>
</li>

</ol>
</div>
