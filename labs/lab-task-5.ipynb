<h3 style="font-family: Times New Roman"><strong>Laboratory Task 5</strong></h3>

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>1. Perform standard imports</b></p>

```python
import torch
import numpy as np
```

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>2. Create a function called <tt>set_seed()</tt> that accepts <tt>seed: int</tt> and returns nothing</b></p>

```python
def set_seed(seed: int) -> None:
    np.random.seed(seed)
    torch.manual_seed(seed)
```

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>3. Create a NumPy array called "arr" with 6 random integers in [0, 5), using seed 42</b></p>

```python
set_seed(42)
arr = np.random.randint(0, 5, size=6).astype('int32')

assert np.array_equal(arr, np.array([3, 4, 2, 4, 4, 1])), "Oops"
print(arr)
```

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>4. Create a tensor "x" from the array above</b></p>

```python
x = torch.from_numpy(arr)
print(x)
```

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>5. Change the dtype of x from int32 to int64</b></p>

```python
print('Old:', x.type())

x = x.type(torch.int64)

print('New:', x.type())
```

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>6. Reshape x into a 3x2 tensor</b></p>

```python
x = x.reshape(3, 2)
print(x)
print(x.shape)
```

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>7. Return the right-hand column of tensor x</b></p>

```python
print(x[:, 1:])
```

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>8. Without changing x, return a tensor of square values of x</b></p>

```python
squares = x.pow(2)
print(squares)
print(x)
```

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>9. Create a tensor y with the same number of elements as x that can be matrix-multiplied with x</b></p>

```python
set_seed(42)
y = torch.randint(0, 5, (2, 3))

assert torch.equal(y, torch.tensor([[2, 2, 1], [4, 1, 0]])), "Oops"
print(y)
print(y.shape, y.numel() == x.numel())
```

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>10. Find the matrix product of x and y</b></p>

```python
prod = torch.mm(x, y)
print(prod)
print(prod.shape)
```

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>Interpretation:</b> <tt>x</tt> has shape (3, 2) and <tt>y</tt> has shape (2, 3). Matrix multiplication needs the columns of the first tensor to match the rows of the second (2 = 2), and both tensors hold 6 elements, so <tt>y</tt> must be (2, 3). The product has shape (3, 2) x (2, 3) = (3, 3). In step 8, <tt>pow(2)</tt> returns a new tensor, so <tt>x</tt> keeps its original values; the in-place version <tt>pow_(2)</tt> would have changed it.</p>
