# Laboratory Task 2: Forward Pass and Error by Hand

<h3 style="font-family: Times New Roman"><strong>Laboratory Task 2</strong></h3>

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>Instruction:</b> Perform a single forward pass and compute for the error.</p>

<p style="font-family:Times New Roman; text-align:justify; font-size:15px">Given: $x = [1, 0, 1]^T$, $y = 1$, activation $f = \max(0, Z_n)$ (ReLU), and the weights and biases from the task. Using the same indexing as the lecture formulas:</p>

$$Z_1 = w_{11}x_1 + w_{13}x_2 + w_{15}x_3 + \theta_1 \qquad Z_2 = w_{12}x_1 + w_{14}x_2 + w_{16}x_3 + \theta_2$$

**Step 1: Hidden layer weighted sums**

$$Z_1 = (0.2)(1) + (0.4)(0) + (-0.5)(1) + (-0.4) = 0.2 + 0 - 0.5 - 0.4 = -0.7$$
$$Z_2 = (-0.3)(1) + (0.1)(0) + (0.2)(1) + (0.2) = -0.3 + 0 + 0.2 + 0.2 = 0.1$$

**Step 2: Apply ReLU to the hidden units**

$$H_1 = \max(0, -0.7) = 0 \qquad H_2 = \max(0, 0.1) = 0.1$$

**Step 3: Output layer weighted sum**

$$Z_3 = w_{21}H_1 + w_{22}H_2 + \theta_3 = (-0.3)(0) + (-0.2)(0.1) + 0.1 = -0.02 + 0.1 = 0.08$$

**Step 4: Apply ReLU to the output**

$$\hat{y} = \max(0, 0.08) = 0.08$$

**Step 5: Error**

$$\delta = \hat{y} - y = 0.08 - 1 = -0.92$$
$$E = \tfrac{1}{2}(\hat{y} - y)^2 = \tfrac{1}{2}(-0.92)^2 = 0.4232 \qquad \text{(squared error } (\hat{y}-y)^2 = 0.8464\text{)}$$

<p style="font-family:Times New Roman; text-align:justify; font-size:15px"><b>Answer:</b> $\hat{y} = 0.08$ and the error is $\hat{y} - y = -0.92$ (the prediction is far below the target of 1).</p>

<p style="font-family:Times New Roman; text-align:justify; font-size:15px">Quick numerical check of the hand computation:</p>

```python
import numpy as np

x = np.array([1, 0, 1])
y = np.array([1])

relu = lambda z: np.maximum(0, z)

W_h = np.array([[ 0.2, -0.3],
                [ 0.4,  0.1],
                [-0.5,  0.2]])
W_o = np.array([[-0.3],
                [-0.2]])
theta_h = np.array([-0.4, 0.2])
theta_o = np.array([0.1])

Z_h = x @ W_h + theta_h
H   = relu(Z_h)
Z_o = H @ W_o + theta_o
y_hat = relu(Z_o)

print("Z1, Z2 :", Z_h)
print("H1, H2 :", H)
print("Z3     :", Z_o)
print("y_hat  :", y_hat)
print("error (y_hat - y)      :", y_hat - y)
print("0.5 * (y_hat - y)^2    :", 0.5 * (y_hat - y) ** 2)
```
